```python
def corrida_para_o_100():
    print("1 - Computador joga em primeiro")
    print("2 - Computador joga em segundo")
    
    modalidade = input("Escolha a modalidade (1 ou 2): ")
    total = 0
    
   
    def jogada_computador(total):
        resto = total% 11
        if resto == 0:
            jogada = 1
        elif resto == 1:
            jogada = 1 
        else:
            jogada = (12 - resto) if resto > 1 else 1
            
       
        jogada = max(1, min(10, jogada))
        
        if total + jogada > 100:
            jogada = 100 - total
            
        return jogada

    def jogada_utilizador(total):
        while True:
            try:
                jogada = int(input(f"Total: {total}. Quantos somas (1 a 10)? "))
                if 1 <= jogada <= 10 and (total + jogada <= 100):
                    return jogada
                print("Jogada inválida. Escolhe um número entre 1 e 10 que não ultrapasse 100.")
            except ValueError:
                print("Entrada inválida. Digita um número inteiro.")


    turno_computador = (modalidade == "1")

    while total < 100:
        if turno_computador:
            jogada = jogada_computador(total)
            total += jogada
            print(f"\n-> O Computador somou {jogada}. Total atual: {total}")
            
            if total == 100:
                print("\nO Computador ganhou!")
                break
            turno_computador = False
        else:
            jogada = jogada_utilizador(total)
            total += jogada
            print(f"-> Tu somaste {jogada}. Total atual: {total}\n")
            
            if total == 100:
                print("\nVenceste o jogo!")
                break
            turno_computador = True

if __name__ == "__main__":
    corrida_para_o_100()
```
